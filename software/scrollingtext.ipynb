```python
import tkinter as tk
import random

class MarketTicker:
    def __init__(self, root):
        self.root = root
        self.root.title("Market Price Ticker")
        # Set window size and background color
        self.root.geometry("1000x120") 
        self.root.configure(bg="black")

        # Create a canvas for smooth scrolling
        self.canvas = tk.Canvas(root, bg="white", height=120, width=1000, highlightthickness=0)
        self.canvas.pack()

        # List of normal, real market symbols
        self.symbols = ["AAPL", "MSFT", "NVDA", "TSLA"]

        # Create the text object on the canvas using Arial font
        self.text_id = self.canvas.create_text(
            1000, 60, 
            text="", 
            fill="#000000", # Bright terminal green
            font=("Arial", 28, "bold"), 
            anchor="w"
        )
        
        self.generate_new_text()
        self.scroll()

    def generate_new_text(self):
        """Generates a new string of market prices."""
        items = []
        
        # Generate 8 random symbol/price pairs for the ticker stream
        for _ in range(8):
            sym = random.choice(self.symbols)
            
            # Simulate a random stock price
            price = round(random.uniform(100, 600), 2)
            
            # Simple, clean formatting (e.g., MSFT : $340.50)
            feed_string = f'{sym}  {price:.2f}'
            items.append(feed_string)
            
        # Join them together with a diamond separator
        self.current_text = "            ".join(items) + "           "
        self.canvas.itemconfig(self.text_id, text=self.current_text)

    def scroll(self):
        """Moves the text to the left continuously."""
        # Move text 3 pixels to the left
        self.canvas.move(self.text_id, -3, 0)
        
        # Check if the text has completely scrolled off the left edge of the screen
        bbox = self.canvas.bbox(self.text_id)
        if bbox and bbox[2] < 0:
            # Reset position to the far right edge
            self.canvas.coords(self.text_id, 1000, 60)
            # Generate new data for the next scroll loop
            self.generate_new_text()
            
        # Loop this function every 16 milliseconds (~60 frames per second)
        self.root.after(40, self.scroll)

if __name__ == "__main__":
    root = tk.Tk()
    app = MarketTicker(root)
    root.mainloop()
```
